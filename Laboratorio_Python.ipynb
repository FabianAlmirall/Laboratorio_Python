## 2. Tipos de estructuras de datos

### Tuplas

Colecciones **ordenadas** e **inmutables**: una vez creadas, no se pueden modificar sus elementos.

````python
coordenada = (9.93, -84.08)  # (latitud, longitud)
mixta = (1, "dos", 3.0, True)

print(coordenada[0])   # acceso por índice -> 9.93

# desempaquetado de tuplas
lat, lon = coordenada
print(lat, lon)

# intento de modificar una tupla lanza error:
# coordenada[0] = 10  -> TypeError
````

### Listas

Colecciones **ordenadas** y **mutables**. Es la estructura de propósito general más usada en Python.

````python
notas = [85, 90, 78, 92]

# indexación: empieza en 0; los índices negativos cuentan desde el final
print(notas[0])   # 85 (primer elemento)
print(notas[-1])  # 92 (último elemento)

# slicing (rebanado): lista[inicio:fin:paso], "fin" no se incluye
print(notas[1:3])   # [90, 78]
print(notas[:2])    # [85, 90]
print(notas[::-1])  # [92, 78, 90, 85] -> lista invertida

# métodos principales
notas.append(88)        # agrega un elemento al final -> [85, 90, 78, 92, 88]
notas.extend([70, 60])  # agrega varios elementos al final
ultimo = notas.pop()    # elimina y retorna el último elemento
notas.pop(0)            # elimina el elemento en la posición 0

print(notas)
print(ultimo)

# otras operaciones útiles
print(len(notas))       # cantidad de elementos
print(sorted(notas))    # lista ordenada (no modifica la original)
print(sum(notas) / len(notas))  # promedio
````

### Diccionario

Colecciones de pares **clave-valor**, no ordenadas por posición sino por clave (desde Python 3.7 conservan el orden de inserción).

````python
estudiante = {
    "nombre": "Ricardo",
    "carrera": "Estadística",
    "promedio": 67.4
}

print(estudiante["nombre"])      # acceso por clave
print(estudiante.get("edad"))    # None, no lanza error si la clave no existe

# agregar o modificar
estudiante["edad"] = 18
estudiante["promedio"] = 87.0

# recorrer un diccionario
for clave, valor in estudiante.items():
    print(clave, "->", valor)

print(estudiante.keys())    # todas las claves
print(estudiante.values())  # todos los valores
````

### Conjuntos (Sets)

Colecciones **no ordenadas** de elementos **únicos** (no permiten duplicados). Útiles para operaciones de teoría de conjuntos.

````python
cursos_Ricardo = {"Análisis Multivariado y Mineria de Datos", "Diseño de Experimentos", "Modelos Lineales Avanzados"}
cursos_Shu = {"Teoría Estadística", "Series de Tiempo", "Modelos Probabilísticos I"}

cursos_Ricardo.add("Modelos de Regresión Aplicados")  # agregar elemento al conjunto

print(cursos_Ricardo | cursos_Shu)  # unión de dos conjuntos
print(cursos_Ricardo & cursos_Shu)  # intersección de conjuntos
print(cursos_Ricardo - cursos_Shu)  # diferencia (solo en Ricardo)

# eliminar duplicados de una lista usando un conjunto (set)
numeros = [1, 2, 2, 3, 3, 3, 4]
print(set(numeros))  # {1, 2, 3, 4}
````

## 4. Ciclos y bucles

### `for`

Recorre los elementos de una secuencia (lista, tupla, string, rango, etc.) directamente, sin necesidad de un contador manual.

````python
# recorrer una lista
for curso in ["Inferencia Estadística", "Inglés para Estadística II", "Demografía y Biostadística"]:
    print(curso)

# range(inicio, fin, paso): "fin" no se incluye
for i in range(0, 10, 2):
    print(i)  # 0, 2, 4, 6, 8

# enumerate() da el índice y el valor a la vez
for i, curso in enumerate(["Análisis Exploratorio de Datos", "Teoría Estadística"]):
    print(i, curso)

# comprensión de listas: forma compacta de construir una lista con un for
cuadrados = [n**2 for n in range(5)]
print(cuadrados)  # [0, 1, 4, 9, 16]
````

### `while`

Repite un bloque mientras una condición sea verdadera. Requiere actualizar manualmente la variable de control para evitar un ciclo infinito.

````python
contador = 0

while contador < 5:
    print(f"Iteración {contador}")
    contador += 1  # equivalente a contador = contador + 1

# while con break y continue
n = 0
while True:
    n += 1
    if n % 2 == 0:
        continue  # salta al siguiente ciclo sin ejecutar lo que sigue
    if n > 7:
        break     # rompe el ciclo
    print(n)  # imprime 1, 3, 5, 7
````

## 5. Simulación similar a la del semestre pasado

````python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
from scipy.stats import chi2

# 1. Paramétros iniciales
r = np.array([6, 7, 5, 8])
n = np.sum(r)

dosis = np.repeat([1, 2, 3, 4], r)

beta = np.array([-2.4, np.log(1.5 / 2), np.log(1.5 / 2), np.log(1.5)])
tau_4 = -np.sum(beta[1:])  # los objetos en Python comienzan en 0, por lo que beta[1:] equivale a beta[1:4]
PL = beta[0] + np.append(beta[1:], tau_4)

PL1 = PL[dosis - 1]
pi = np.exp(PL1) / (1 + np.exp(PL1))

p_values = []
np.random.seed(42)  # semilla para garantizar la reproducibilidad de los resultados

# 2. Bucle de simulación
for i in range(1000):
    tot = np.random.randint(20, 26, size=n)
    malas = np.random.binomial(tot, pi)
    buenas = tot - malas

    df = pd.DataFrame({"malas": malas, "buenas": buenas, "tot": tot, "dosis": dosis})

    # Calcular la proporción de respuestas
    df['prop'] = df['malas'] / df['tot']

    # Ajustar GLM usando la proporción y pasando 'tot' en var_weights
    mod = smf.glm(
        formula="prop ~ C(dosis)",
        data=df,
        family=sm.families.Binomial(),
        var_weights=df['tot']  # indica que no todas las filas tienen el mismo número de ensayos
    ).fit()

    # Modelo nulo para la prueba de Razón de Verosimilitud (LRT)
    mod_null = smf.glm(
        formula="prop ~ 1",
        data=df,
        family=sm.families.Binomial(),
        var_weights=df['tot']
    ).fit()

    # Prueba LRT a mano
    lr_stat = mod_null.deviance - mod.deviance
    df_diff = mod_null.df_resid - mod.df_resid
    p_val = chi2.sf(lr_stat, df_diff)

    p_values.append(p_val)

# 3. Cálculo de la potencia
pot = np.mean(np.array(p_values) < 0.05)

print(f"La potencia para esta simulación es de: {pot:.4f}")
````

## 6. Álgebra Lineal y Computación Numérica con NumPy

### Creación y manipulación de vectores y matrices

````python
import numpy as np

# vector (arreglo 1D)
v = np.array([1, 2, 3])

# matriz (arreglo 2D)
M = np.array([[1, 2], [3, 4]])

print(v.shape)   # (3,)
print(M.shape)   # (2, 2)

# funciones de creación
ceros = np.zeros((2, 3))       # matriz 2x3 de ceros
unos = np.ones((3, 2))         # matriz 3x2 de unos
identidad = np.eye(3)          # matriz identidad 3x3
secuencia = np.arange(0, 10, 2)  # [0, 2, 4, 6, 8]
aleatorios = np.random.normal(0, 1, size=(2, 2))  # matriz 2x2 ~ N(0,1)

# indexación y slicing (igual que en listas, pero con soporte multidimensional)
print(M[0, 1])    # elemento fila 0, columna 1 -> 2
print(M[:, 0])    # toda la primera columna -> [1, 3]
print(M[0, :])    # toda la primera fila -> [1, 2]

# reshape: cambia la forma de un arreglo sin cambiar sus datos
r = np.arange(6).reshape(2, 3)
print(r)
````

### Operaciones matriciales para análisis multivariado

````python
A = np.array([[2, 0], [0, 3]])
B = np.array([[1, 1], [1, 1]])

# operaciones elemento a elemento
print(A + B)
print(A * B)   # multiplicación elemento a elemento, NO multiplicación matricial

# multiplicación matricial real: @ o np.dot()
print(A @ B)
print(np.dot(A, B))

# transpuesta
print(A.T)

# determinante e inversa (submódulo linalg)
print(np.linalg.det(A))    # determinante
print(np.linalg.inv(A))    # matriz inversa

# valores y vectores propios (autovalores/autovectores), clave en componentes principales
valores, vectores = np.linalg.eig(A)
print(valores)
print(vectores)

# matriz de covarianza a partir de datos (variables en filas por defecto en rowvar=True)
datos = np.random.normal(0, 1, size=(3, 100))  # 3 variables, 100 observaciones
cov = np.cov(datos)
print(cov.shape)  # (3, 3)
````
