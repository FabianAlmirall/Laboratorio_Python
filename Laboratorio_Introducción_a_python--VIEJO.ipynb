### Simulación similar a la del semestre pasado

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
from scipy.stats import chi2

# 1. Paramétros iniciales
r = np.array([6, 7, 5, 8])
n = np.sum(r)

dosis = np.repeat([1, 2, 3, 4], r)

beta = np.array([-2.4, np.log(1.5 / 2), np.log(1.5 / 2), np.log(1.5)])
tau_4 = -np.sum(beta[1:]) # Recordar que los objetos en pyhton comenzan en 0, por lo que beta[1:] es equivalente a beta[1:4]
PL = beta[0] + np.append(beta[1:], tau_4)

PL1 = PL[dosis - 1]
pi = np.exp(PL1) / (1 + np.exp(PL1))

p_values = []
np.random.seed(42) # Se utliza una semila para garantizar la reproducibilidad de los resultados

# 2. Bucle de simulación
for i in range(1000):
    tot = np.random.randint(20, 26, size=n)
    malas = np.random.binomial(tot, pi)
    buenas = tot - malas

    df = pd.DataFrame({"malas": malas, "buenas": buenas, "tot": tot, "dosis": dosis})
    
    # Calcular la proporción de respuestas
    df['prop'] = df['malas'] / df['tot']

    # Ajustar GLM usando la proporción y pasando 'tot' en var_weights
    mod = smf.glm(
        formula="prop ~ C(dosis)", 
        data=df, 
        family=sm.families.Binomial(),
        var_weights=df['tot'] # Se le indica a Python que no todas las filas tienen el mismo número de observaciones y le especifica la cantidad de ensayos
    ).fit()

    # Modelo nulo para la prueba de Razón de Verosimilitud (LRT)
    mod_null = smf.glm(
        formula="prop ~ 1", 
        data=df, 
        family=sm.families.Binomial(),
        var_weights=df['tot']
    ).fit()

    # Prueba LRT a mano
    lr_stat = mod_null.deviance - mod.deviance
    df_diff = mod_null.df_resid - mod.df_resid
    p_val = chi2.sf(lr_stat, df_diff)

    p_values.append(p_val)

# 3. Cálculo de la potencia
pot = np.mean(np.array(p_values) < 0.05)

print(f"La potencia para esta simulación es de: {pot:.4f}")
```
